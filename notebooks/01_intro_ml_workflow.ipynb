## ۱. یادگیری ماشین چیست؟

در **برنامه‌نویسی سنتی**، ما قانون‌ها را خودمان می‌نویسیم: «اگر ایمیل شامل کلمه‌ی «جایزه» بود، آن را هرزنامه در نظر بگیر». ورودی برنامه، داده و قانون است و خروجی آن، جواب.

در **یادگیری ماشین (Machine Learning)**، قانون را نمی‌نویسیم. به‌جای آن، تعداد زیادی مثال همراه با جواب درست به الگوریتم می‌دهیم و الگوریتم خودش قانون (یعنی **مدل**) را از روی داده یاد می‌گیرد. سپس از این مدل برای پیش‌بینی جواب مثال‌های جدید استفاده می‌کنیم.

تعریف معروف تام میچل (Tom Mitchell): برنامه‌ای که با کسب **تجربه‌ی E** در انجام **وظیفه‌ی T**، طبق **معیار عملکرد P** بهتر می‌شود، در حال یادگیری است. برای فیلتر هرزنامه: T تشخیص هرزنامه است، E ایمیل‌هایی که کاربران برچسب زده‌اند و P درصد ایمیل‌هایی که درست تشخیص داده می‌شوند.

یادگیری ماشین وقتی مفید است که:
- نوشتن قانون به‌صورت دستی سخت یا غیرممکن باشد (مثل تشخیص چهره)
- قانون‌ها مدام تغییر کنند (مثل الگوهای تقلب بانکی)
- داده‌ی زیادی در دسترس باشد و بخواهیم الگوهای پنهان آن را کشف کنیم

</div>

<div dir="rtl">

## ۲. انواع یادگیری ماشین

| نوع | داده‌ی آموزشی | هدف | مثال |
|---|---|---|---|
| **نظارت‌شده (Supervised)** | ویژگی‌ها + برچسب درست | پیش‌بینی برچسب داده‌ی جدید | پیش‌بینی قیمت خانه، تشخیص هرزنامه |
| **بدون نظارت (Unsupervised)** | فقط ویژگی‌ها، بدون برچسب | کشف ساختار پنهان داده | بخش‌بندی مشتریان، کاهش ابعاد |
| **تقویتی (Reinforcement)** | پاداش یا جریمه از محیط | یادگیری بهترین رفتار | بازی‌های رایانه‌ای، رباتیک |

یادگیری نظارت‌شده خودش دو دسته دارد:
- **رگرسیون (Regression):** خروجی یک عدد پیوسته است؛ مثل قیمت خانه یا دمای فردا.
- **طبقه‌بندی (Classification):** خروجی یک دسته از میان چند دسته‌ی مشخص است؛ مثل «هرزنامه / عادی» یا نوع گل.

در این دوره بیشتر روی یادگیری نظارت‌شده و بدون نظارت تمرکز می‌کنیم و در پایان سراغ شبکه‌های عصبی می‌رویم.

</div>

<div dir="rtl">

## ۳. واژگان پایه

- **مجموعه‌داده (Dataset):** جدولی از داده‌ها
- **نمونه (Sample / Instance):** هر سطر جدول؛ مثلاً یک گل یا یک مشتری
- **ویژگی (Feature):** هر ستونی که برای پیش‌بینی استفاده می‌شود؛ معمولاً ماتریس ویژگی‌ها را با $X$ نشان می‌دهیم
- **برچسب یا هدف (Label / Target):** ستونی که می‌خواهیم پیش‌بینی کنیم؛ معمولاً با $y$ نشان داده می‌شود
- **مدل (Model):** تابعی که از روی $X$ مقدار $y$ را تخمین می‌زند
- **آموزش (Training / Fitting):** فرایند یافتن بهترین پارامترهای مدل از روی داده
- **تعمیم (Generalization):** توانایی مدل در پیش‌بینی درست برای داده‌هایی که هرگز ندیده است. **هدف اصلی یادگیری ماشین همین است**، نه حفظ کردن داده‌ی آموزشی.

</div>

<div dir="rtl">

## ۴. جریان کار یک پروژه‌ی یادگیری ماشین

تقریباً همه‌ی پروژه‌های یادگیری ماشین این مراحل را طی می‌کنند:

1. **تعریف مسئله:** چه چیزی را می‌خواهیم پیش‌بینی کنیم و موفقیت را چطور می‌سنجیم؟
2. **جمع‌آوری و بارگذاری داده**
3. **تحلیل اکتشافی داده (EDA):** شناخت داده، توزیع‌ها، مقادیر گمشده و روابط
4. **پیش‌پردازش:** پاک‌سازی، تبدیل ویژگی‌های متنی به عدد، مقیاس‌بندی
5. **تقسیم داده به آموزش و آزمون**
6. **آموزش مدل**
7. **ارزیابی مدل روی داده‌ی آزمون**
8. **بهبود:** امتحان مدل‌ها و تنظیمات دیگر
9. **استقرار (Deployment):** استفاده از مدل در دنیای واقعی

در ادامه‌ی همین جلسه، این مراحل را روی یک مجموعه‌داده‌ی کوچک و معروف اجرا می‌کنیم. در جلسه‌های بعد، هر مرحله را عمیق‌تر یاد می‌گیریم.

</div>

## ۶. مثال عملی: طبقه‌بندی گل زنبق (Iris)

### ۶.۱ تعریف مسئله

مجموعه‌داده‌ی **Iris** شامل اندازه‌های ۱۵۰ گل زنبق از سه گونه است: *setosa*، *versicolor* و *virginica*. برای هر گل چهار ویژگی اندازه‌گیری شده است: طول و عرض کاسبرگ (sepal) و طول و عرض گلبرگ (petal).

**مسئله:** با داشتن این چهار اندازه، گونه‌ی گل را پیش‌بینی کنیم. این یک مسئله‌ی **طبقه‌بندی** با سه کلاس است. معیار موفقیت را **دقت (accuracy)** در نظر می‌گیریم: درصد گل‌هایی که گونه‌ی آن‌ها درست پیش‌بینی می‌شود.

### ۶.۲ بارگذاری داده

این مجموعه‌داده همراه scikit-learn ارائه می‌شود و نیازی به دانلود ندارد.

</div>

```python
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)
df = iris.frame
df["species"] = df["target"].map(dict(enumerate(iris.target_names)))
df.head()
```

### ۶.۳ تحلیل اکتشافی داده

اولین قدم همیشه شناختن داده است: چند نمونه داریم؟ نوع ستون‌ها چیست؟ مقدار گمشده داریم؟ کلاس‌ها متوازن‌اند؟

</div>

```python
print("Shape:", df.shape)
print()
df.info()
```

```python
df.describe().round(2)
```

```python
df["species"].value_counts()
```

<div dir="rtl">

داده ۱۵۰ نمونه و ۴ ویژگی عددی دارد، هیچ مقدار گمشده‌ای ندارد و از هر گونه دقیقاً ۵۰ نمونه موجود است؛ یعنی کلاس‌ها **متوازن** هستند.

حالا با یک نمودار ببینیم ویژگی‌ها چقدر می‌توانند گونه‌ها را از هم جدا کنند:

</div>

```python
sns.pairplot(df.drop(columns="target"), hue="species", height=2.0, corner=True)
plt.show()
```

<div dir="rtl">

از نمودار پیداست که گونه‌ی *setosa* با ویژگی‌های گلبرگ (petal) کاملاً از بقیه جداست. دو گونه‌ی دیگر کمی هم‌پوشانی دارند. پس انتظار داریم یک مدل ساده هم دقت بالایی داشته باشد، ولی احتمالاً بین *versicolor* و *virginica* چند اشتباه رخ دهد.

</div>

<div dir="rtl">

### ۶.۴ آماده‌سازی داده و تقسیم به آموزش و آزمون

ماتریس ویژگی‌ها $X$ و بردار برچسب $y$ را جدا می‌کنیم. سپس داده را به دو بخش تقسیم می‌کنیم:

- **داده‌ی آموزش (Train):** مدل فقط این بخش را می‌بیند و از آن یاد می‌گیرد.
- **داده‌ی آزمون (Test):** کنار گذاشته می‌شود تا در پایان، عملکرد مدل روی داده‌ی **نادیده** سنجیده شود.

چرا؟ چون اگر مدل را روی همان داده‌ای بسنجیم که با آن آموزش دیده، مثل این است که سؤال‌های امتحان را قبلاً به دانشجو داده باشیم؛ نمره‌ی بالا هیچ چیزی درباره‌ی یادگیری واقعی نمی‌گوید.

پارامتر `stratify=y` باعث می‌شود نسبت کلاس‌ها در هر دو بخش یکسان بماند و `random_state` تقسیم را تکرارپذیر می‌کند.

</div>

```python
from sklearn.model_selection import train_test_split

X = df[iris.feature_names]
y = df["species"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, stratify=y, random_state=42
)
print("Train:", X_train.shape, " Test:", X_test.shape)
```

<div dir="rtl">

### ۶.۵ مدل پایه (Baseline)

قبل از ساختن مدل واقعی، همیشه یک **مدل پایه** بسازید: ساده‌ترین پیش‌بینی ممکن. اگر مدل ما از مدل پایه بهتر نباشد، هیچ چیزی یاد نگرفته است. اینجا مدل پایه همیشه پرتکرارترین کلاس را پیش‌بینی می‌کند.

</div>

```python
from sklearn.dummy import DummyClassifier

baseline = DummyClassifier(strategy="most_frequent")
baseline.fit(X_train, y_train)
print(f"Baseline accuracy: {baseline.score(X_test, y_test):.3f}")
```

<div dir="rtl">

دقت حدود ۳۳٪ است، یعنی همان شانس حدس تصادفی بین سه کلاس متوازن.

### ۶.۶ آموزش اولین مدل: k نزدیک‌ترین همسایه (k-NN)

ایده‌ی **k-NN** بسیار ساده است: برای پیش‌بینی گونه‌ی یک گل جدید، $k$ گل شبیه به آن را در داده‌ی آموزش پیدا کن و گونه‌ای را انتخاب کن که بین آن‌ها بیشتر تکرار شده است. جزئیات این الگوریتم را در جلسه‌ی ۶ می‌بینیم.

همه‌ی مدل‌های scikit-learn رابط یکسانی دارند:
- `model.fit(X_train, y_train)` برای آموزش
- `model.predict(X_new)` برای پیش‌بینی
- `model.score(X_test, y_test)` برای ارزیابی سریع

</div>

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)

y_pred = knn.predict(X_test)
print(y_pred[:10].tolist())
```

<div dir="rtl">

### ۶.۷ ارزیابی مدل

**دقت (Accuracy)** نسبت پیش‌بینی‌های درست به کل پیش‌بینی‌هاست. **ماتریس درهم‌ریختگی (Confusion Matrix)** نشان می‌دهد هر کلاس با کدام کلاس اشتباه گرفته شده است: سطرها کلاس واقعی و ستون‌ها کلاس پیش‌بینی‌شده‌اند.

</div>

```python
from sklearn.metrics import accuracy_score, ConfusionMatrixDisplay

print(f"k-NN accuracy: {accuracy_score(y_test, y_pred):.3f}")

ConfusionMatrixDisplay.from_predictions(y_test, y_pred, cmap="Blues")
plt.title("k-NN on Iris (test set)")
plt.show()
```

<div dir="rtl">

همان‌طور که از نمودار اکتشافی انتظار داشتیم، *setosa* بدون خطا تشخیص داده شده و اشتباه‌های احتمالی بین *versicolor* و *virginica* است. مدل ما بسیار بهتر از مدل پایه‌ی ۳۳٪ عمل می‌کند.

### ۶.۸ پیش‌بینی برای یک گل جدید

فرض کنید گلی با این اندازه‌ها پیدا کرده‌ایم. مدل آموزش‌دیده گونه‌ی آن را پیش‌بینی می‌کند:

</div>

```python
new_flower = pd.DataFrame([[5.9, 3.0, 5.1, 1.8]], columns=iris.feature_names)
print("Predicted species:", knn.predict(new_flower)[0])
probs = knn.predict_proba(new_flower)[0]
for species, p in zip(knn.classes_, probs):
    print(f"  P({species}) = {p:.2f}")
```

<div dir="rtl">

## ۷. چرا داده‌ی آزمون جداگانه لازم است؟ نگاهی به بیش‌برازش

مدل k-NN را با $k=1$ امتحان می‌کنیم. در این حالت مدل هر نمونه‌ی آموزشی را دقیقاً «حفظ» می‌کند، چون نزدیک‌ترین همسایه‌ی هر نقطه خودش است.

</div>

```python
for k in [1, 5, 15, 50]:
    model = KNeighborsClassifier(n_neighbors=k).fit(X_train, y_train)
    print(f"k={k:>2}  train accuracy={model.score(X_train, y_train):.3f}  "
          f"test accuracy={model.score(X_test, y_test):.3f}")
```

<div dir="rtl">

با $k=1$ دقت روی داده‌ی آموزش ۱۰۰٪ است، ولی این عدد گمراه‌کننده است؛ مدل فقط داده را حفظ کرده است. معیار واقعی، دقت روی داده‌ی **آزمون** است.

- اگر مدل بیش از حد پیچیده باشد و جزئیات و نویز داده‌ی آموزش را حفظ کند، دچار **بیش‌برازش (Overfitting)** می‌شود.
- اگر بیش از حد ساده باشد (مثل $k=50$ که تقریباً یک‌سوم کل داده‌ی آموزش را میانگین می‌گیرد)، دچار **کم‌برازش (Underfitting)** می‌شود.

پیدا کردن تعادل بین این دو، یکی از مهم‌ترین موضوع‌های کل این دوره است و در جلسه‌ی ۴ مفصل به آن می‌پردازیم.

</div>
